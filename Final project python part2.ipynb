##### RFM Segmentation

```python
df_backup = df.copy()
```

```python
df['Date'] = df['checkin_date'].dt.date
df['Year'] = df['checkin_date'].dt.year
df['Month'] = df['checkin_date'].dt.month
df['Month name'] = df['checkin_date'].dt.month_name()
df['Day'] = df['checkin_date'].dt.day
df['Day name'] = df['checkin_date'].dt.day_name()
```

```python
df = df.sort_values(["Year", "Month"])
```

```python
analysis_date = df["checkin_date"].max() + pd.Timedelta(days=1)

RFM = df.groupby("guest_email").agg(
    r_recency=("checkin_date", lambda x: (analysis_date - x.max()).days),
    f_frequency=("booking_id", "nunique"),
    m_monetary=("total_price_usd", "sum")
).reset_index()
```

```python
RFM
```

```python
RFM["R"] = pd.qcut(RFM["r_recency"].rank(method="first", ascending=False), 5, labels=[1, 2, 3, 4, 5]).astype(int)
RFM["F"] = pd.qcut(RFM["f_frequency"].rank(method="first"), 5, labels=[1, 2, 3, 4, 5]).astype(int)
RFM["M"] = pd.qcut(RFM["m_monetary"].rank(method="first"), 5, labels=[1, 2, 3, 4, 5]).astype(int)
```

```python
RFM
```

```python
RFM["RFM Score"] = (
    RFM['R'].astype(str) +
    RFM['F'].astype(str) +
    RFM['M'].astype(str)
).astype(int)

RFM
```

```python
def segmantation(score):
    score = int(score)
    if score <= 155:
        return "bronze"
    elif score <= 255:
        return "silver"
    elif score <= 355:
        return "gold"
    elif score <= 455:
        return " prem"
    else:
        return "tait"

RFM["Customer segmantation"] = RFM["RFM Score"].apply(segmantation)

RFM
```

```python
RFM.to_csv("hotel_bookings_RFM.csv", index=False)
```
